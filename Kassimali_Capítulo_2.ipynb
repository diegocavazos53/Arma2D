```python
import numpy as np

# 2.1

# Crear dos matrices de 3x3
matriz_A = np.array([[9, -3,  2], [ 3, -5, 4], [-2, -4,  6]])
matriz_B = np.array([[8, -7, -1], [-7,  9, 0], [-1,  0, -6]])

# Sumar las matrices
matriz_C = matriz_A + matriz_B
matriz_D = matriz_A - matriz_B

print("Matriz A:")
print(matriz_A)
print("\nMatriz B:")
print(matriz_B)
print("\nResultado de la suma:")
print(matriz_C)
print("\nResultado de la resta:")
print(matriz_D)
```

```python
# 2.2

# Crear dos matrices de 4x3
matriz_A = np.array([[6, -1, -8, -4], [-8, -4,  3, 6], [-5, -2, -3,  7]])
matriz_B = np.array([[5, -2,  4,  1], [ 4,  5, -6, 7], [-1,  2,  8, -5]])

# Multiplicar y sumar las matrices
# El producto de un escalar por una matriz se hace con el operador '*'
matriz_C = 2*matriz_A + matriz_B
matriz_D = matriz_A - 3*matriz_B

print("Matriz A:")
print(matriz_A)
print("\nMatriz B:")
print(matriz_B)
print("\nMatriz C:")
print(matriz_C)
print("\nMatriz D:")
print(matriz_D)
```

```python
# 2.3

# Crear dos "matrices"
matriz_A = np.array([[-5], [2], [-8], [4]])
matriz_B = np.array([[3, -7, 1, -6]])
# Aunque sea un renglón, lleva los paréntesis rectos dobles para ser matriz.

# Producto punto se lleva a cabo con el operador '@'
matriz_C = matriz_A @ matriz_B
matriz_D = matriz_B @ matriz_A
print("Matriz C:")
print(matriz_C)
print("Matriz D:")
print(matriz_D)
```

```python
# 2.8

# Crear dos matrices
matriz_A = np.array([[14, -5, 11], [-3, 18, -9]])
matriz_B = np.array([[12, -7, 6, 0], [2, 9, 15, -13], [10, -4, -5, 8]])

# Demostración
# La transpuesta de una matriz se obtiene con .T o con la función np.transpose()
matriz_C = (matriz_A @ matriz_B).T
matriz_D = matriz_B.T @ matriz_A.T
matriz_D2 = np.dot(np.transpose(matriz_B), np.transpose(matriz_A))

print("Matriz C (AB)^T:")
print(matriz_C)
print("Matriz D B^T*A^T:")
print(matriz_D)
print("Matriz D2 B^T*A^T:")
print(matriz_D2)
print("Lo que demuestra que (AB)^T = B^T * A^T")
```

```python
# 2.9

# Crear tres matrices
matriz_A = np.array([[-9, 0], [13, 20], [8, -3], [-11, -5]])
matriz_B = np.array([[15, -1, -4], [6, 16, 9]])
matriz_C = np.array([[-7, 10, 6, 0], [-1, 2, -8, -2], [16, 12, 2, 8]])

# Demostración
matriz_D = (matriz_A @ matriz_B @ matriz_C).T
matriz_E = matriz_C.T @ matriz_B.T @ matriz_A.T

print("Matriz D (ABC)^T:")
print("Matriz E C^T * B^T * A^T")
print(matriz_D)
print(matriz_E)
print("Lo que demuestra que (ABC)^T = C^T * B^T * A^T")
```

```python
import sympy as sp
from sympy import symbols, Matrix, sin, cos, tan

# 2.13

# Definimos las variables simbólicas
x = symbols('x')

# Creamos la matriz simbólica
matriz_A = Matrix([[sin(x)**2, -3*x**3, -6*x],
                   [-6*x**2, cos(x)**2, 5],
                   [7, 3*sin(x), tan(x)]])

# Derivamos la matriz simbólica con respecto a x
matriz_B = sp.diff(matriz_A, x)

print("Matriz A:")
display(matriz_A)
print("\nMatriz B:")
display(matriz_B)
```

```python
# 2.14

# Definimos las variables simbólicas
x = symbols('x')

# Creamos las matrices simbólicas
matriz_A = Matrix([[-3*x, 5],
                   [4*x**2, -x**3],
                   [-7, 5*x],
                   [2*x**3, -x**2]])
matriz_B = Matrix([[2*x**2, -x],
                   [-12*x, 8],
                   [2*x**3, -3*x**2],
                   [-1, 6*x]])

# Derivamos con respecto a x
matriz_C = sp.diff(matriz_A+matriz_B, x)

print("d(A+B)/dx:")
display(matriz_C)
```

```python
# 2.15

# Definimos las variables simbólicas
x = symbols('x')

# Creamos las matrices simbólicas
matriz_A = Matrix([[4*x, 2, -5*x**2],
                   [2, -3*x**3, -x],
                   [-5*x**2, -x, 7]])
matriz_B = Matrix([[-5*x**3, -x],
                   [6, -3*x**2],
                   [2*x**2, 4*x]])

# Derivamos con respecto a x
matriz_C = sp.diff(matriz_A @ matriz_B, x)

print("d(AB)/dx:")
display(matriz_C)
```

```python
# 2.16

# Definimos las variables simbólicas
x, y, z = symbols('x y z')

# Creamos la matriz simbólica
matriz_A = Matrix([[x**2, -y**2, -2*z**2],
                   [-y**2, 3*x*y, -y*z],
                   [2*z**2, -y*z, 4*x*z]])

# Derivamos con respecto a x
matriz_C = sp.diff(matriz_A, x)
matriz_D = sp.diff(matriz_A, y)
matriz_E = sp.diff(matriz_A, z)

print("dA/dx:")
display(matriz_C)
print("dA/dy:")
display(matriz_D)
print("dA/dz:")
display(matriz_E)
```

```python
# 2.17

# Definimos las variables simbólicas
x, L = symbols('x L')

# Creamos la matriz simbólica
matriz_A = Matrix([[6*x**2, -x**3, 9],
                   [-x**3, 2*x**4, 4*x],
                   [9, 4*x, 5*x**3]])

# Integramos con respecto a x
matriz_B = sp.integrate(matriz_A, (x, 0, L)) # Tuple

print("int(Adx), 0, L:")
display(matriz_B)
```

```python
# 2.19

# Definimos las variables simbólicas
x, L = symbols('x L')

# Creamos las matrices simbólicas
matriz_A = Matrix([[-x**3, 2*x**2, 3],
                   [2*x, -x**2, 2*x**3]])
matriz_B = Matrix([[-2*x, x**2],
                   [5, -2*x],
                   [3*x**3, -3]])

# Integramos con respecto a x
matriz_C = sp.integrate(matriz_A @ matriz_B, (x, 0, L)) # Tuple

print("int(Adx), 0, L:")
display(matriz_C)
```

```python
# 2.20

# Crear dos matrices
matriz_A = np.array([[-0.28, -0.96, 0, 0],
                     [0.96, -0.28, 0, 0],
                     [0, 0, -0.28, -0.96],
                     [0, 0, 0.96, -0.28]])
matriz_B = np.array([[-0.28, 0.96, 0, 0],
                     [0.96, -0.28, 0, 0],
                     [0, 0, -0.28, 0.96],
                     [0, 0, 0.96, -0.28]])

print("Si A*A^T da la matriz identidad, la matriz es ortogonal.")
matriz_C = matriz_A @ matriz_A.T
matriz_D = matriz_B @ matriz_B.T
print("Matriz C, A*A^T:")
print(matriz_C)
print("Matriz D, B*B^T:")
print(matriz_D)
print("La matriz D no es ortogonal")
```

```python
# 2.21

# Matriz de coeficientes
matriz_coeficientes = np.array([[2, -3, 1],
                                [-9, 5, 3],
                                [4, 7, -8]])
# Vector de resultados
vector_resultados = np.array([-18, 18, 53])
# Solución
solucion = np.linalg.solve(matriz_coeficientes, vector_resultados)

print("\nLa solución del sistema de ecuaciones es:")
print(f"x = {solucion[0]:.2f}")
print(f"y = {solucion[1]:.2f}")
print(f"z = {solucion[2]:.2f}")
```
